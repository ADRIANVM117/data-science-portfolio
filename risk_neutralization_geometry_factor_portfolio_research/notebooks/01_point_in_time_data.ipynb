# Point-in-Time Data Research

Research narrative scaffold only. No SEC/market-data join is implemented here.

## 1. Research question

How does the frozen information set determine whether a fact or price observation is eligible at a portfolio decision?


## 2. Frozen conventions

Decision time: regular-session open on session `t`. Market information: completed observations through `t-1`. SEC fundamentals: first eligible XNYS session strictly after `acceptanceDateTime`. Scheduled early-close sessions are eligible.
